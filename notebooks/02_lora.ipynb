# Stage 2 — Fine-Tuning with LoRA (Low-Rank Adaptation)

In this stage of the project, we apply LoRA (Low-Rank Adaptation) to fine-tune pre-trained language models using the dataset created in Stage 1.

The goal is to adapt both causal language models and sequence-to-sequence models to the domain of washing machine manuals, improving their ability to generate accurate and context-aware responses.

---

We will train:

- 2 causal language models:
  - TinyLlama-1.1B-Chat (`TinyLlama/TinyLlama-1.1B-Chat-v1.0`)
  - Microsoft Phi-2 (`microsoft/phi-2`)

- 2 sequence-to-sequence models:
  - mT5 Small (`google/mt5-small`)
  - BART Base (`facebook/bart-base`)

All models will be evaluated before and after fine-tuning to demonstrate domain adaptation performance improvements.
